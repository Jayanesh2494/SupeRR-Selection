# Phase 1 — Data Exploration and Validation

This notebook profiles the six supplied CSVs and runs key relationship and cricket data-quality checks. It does not modify inputs, calculate rankings, or assign pressure labels.

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "src").is_dir() and (ROOT.parent / "src").is_dir():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from data_exploration import load_datasets, profile_datasets, resolve_data_dir
from data_validation import validate_datasets

data_dir = resolve_data_dir(ROOT)
datasets = load_datasets(data_dir)
print(f"Data directory: {data_dir}")
print({name: frame.shape for name, frame in datasets.items()})
```

## Dataset profiles

The next cell shows each table's inferred schema, missingness, exact duplicates, candidate single-column keys, and sample rows.

```python
profiles = profile_datasets(datasets)
for name, profile in profiles.items():
    print(f"\n{name}: {profile['rows']:,} rows x {profile['column_count']} columns")
    print(f"Exact duplicate rows: {profile['exact_duplicate_rows']}")
    print(f"Non-null unique columns: {profile['non_null_unique_columns']}")
    display(pd.DataFrame(profile['columns']))
    display(pd.DataFrame(profile['sample_records']))
```
